#KPI Analysis

```python
total_sales = df["Sales"].sum()
total_cost = df["Total_Cost"].sum()
total_profit = df["Profit"].sum()

total_orders = df["Order_ID"].nunique()

average_order_value = (
    total_sales / total_orders
    if total_orders != 0
    else 0
)

highest_order = df["Sales"].max()
lowest_order = df["Sales"].min()

average_profit = df["Profit"].mean()
average_profit_margin = df["Profit_Margin"].mean()

print("Total Sales:", total_sales)
print("Total Cost:", total_cost)
print("Total Profit:", total_profit)
print("Average Order Value:", average_order_value)
print("Highest Order:", highest_order)
print("Lowest Order:", lowest_order)
print("Total Orders:", total_orders)
print("Average Profit:", average_profit)
print("Average Profit Margin:", average_profit_margin)
```

#Group Analysis

```python
#Sales by Region
sales_by_region = df.groupby("Region")["Sales"].sum()
print(sales_by_region)
#Profit by Region
profit_by_region = df.groupby("Region")["Profit"].sum()
print(profit_by_region)
#Sales by Product
sales_by_product = df.groupby("Product")["Sales"].sum()
print(sales_by_product)
#Profit by Product
profit_by_product = df.groupby("Product")["Profit"].sum()
print(profit_by_product)
#Sales by Category
sales_by_category = df.groupby("Category")["Sales"].sum()
print(sales_by_category)
#Sales by Channel
sales_by_channel = df.groupby("Sales_Channel")["Sales"].sum()
print(sales_by_channel)
#Salesperson Performance
salesperson_performance = df.groupby("Salesperson")["Sales"].sum().sort_values(ascending=False).head(3)
print(salesperson_performance)
```

#Ranking


```python
#Top 10 orders
top_10_orders = df.nlargest(10,"Sales")
print(top_10_orders)
```

```python
#top 5 products
top_5_products = df.groupby("Product")["Sales"].sum().sort_values(ascending=False).head(5)
print(top_5_products)
#
```

```python
#top sales person
top_salesperson = (
    df.groupby("Salesperson")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(1)
)

print(top_salesperson)
```

```python
#Best region
best_region = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(1)
)

print(best_region)
#Best category
best_category = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(1)
)

print(best_category)
#Most profitable product
most_profitable_product = (
    df.groupby("Product")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .head(1)
)

print(most_profitable_product)
```
